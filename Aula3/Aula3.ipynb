## The Role of Probability in Statistics

### Why Probability?

- **Data Uncertainty**: Observations are subject to errors and noise.
- **Model Uncertainty**: Models are simplifications of reality.
- **Inference**: We use probability to make inferences about the universe from incomplete data.

### Two Interpretations of Probability

1. **Frequentist Probability**
   - Defines probability as the long-run frequency of events.
   - *Example*: The probability of a fair coin landing heads is 0.5 because, over many tosses, half will be heads.

2. **Bayesian Probability**
   - Defines probability as a degree of belief or confidence in an event.
   - *Example*: Before tossing a coin, you might believe there's a 50% chance it will land heads based on prior knowledge.

## Bayesian vs. Frequentist Approaches

### Frequentist Approach

- **Philosophy**
  - Parameters are fixed but unknown quantities.
  - Probability applies only to data, not parameters.
- **Focus**
  - \( P(D|H) \): Probability of observing data \( D \) given hypothesis \( H \).
- **Inference**
  - Relies on sampling distributions and long-term frequencies.
- **Examples**
  - Hypothesis testing.
  - Confidence intervals.

### Bayesian Approach

- **Philosophy**
  - Parameters are random variables with probability distributions.
  - Probability quantifies our belief about parameters.
- **Focus**
  - \( P(H|D) \): Probability of hypothesis \( H \) given data \( D \).
- **Inference**
  - Updates prior beliefs with new data using Bayes' theorem.
- **Examples**
  - Posterior probability distributions.
  - Credible intervals.

### Key Differences

- **Interpretation of Probability**
  - Frequentist: Objective and long-term frequency.
  - Bayesian: Subjective degree of belief.
- **Treatment of Parameters**
  - Frequentist: Fixed but unknown.
  - Bayesian: Random variables with distributions.
- **Use of Prior Information**
  - Frequentist: Typically does not incorporate prior beliefs.
  - Bayesian: Integrates prior knowledge explicitly.

## Bayes' Theorem in Detail

### The Foundation of Bayesian Inference

- **Bayes' Theorem Formula**
  $$
  P(H|D) = \frac{P(H) \cdot P(D|H)}{P(D)}
  $$

### Components Explained

1. **Posterior Probability ($P(H|D)$)**
   - The updated probability of the hypothesis after considering the data.
   - Represents our new belief.

2. **Prior Probability ($P(H)$)**
   - The initial belief about the hypothesis before observing data.
   - Can be based on previous studies or theoretical considerations.

3. **Likelihood ($P(D|H)$)**
   - The probability of observing the data given the hypothesis.
   - Connects the model with the observations.

4. **Evidence ($P(D)$)**
   - The total probability of observing the data under all possible hypotheses.

### Important Notes

- **Posterior ≠ Likelihood**
  - The likelihood is a component of the posterior but not the same.
- **Role of the Evidence**
  - While it normalizes the posterior, it's crucial for comparing different models (model selection).

## Constructing Likelihoods in Cosmology

### Importance of Likelihoods

- **Connecting Theory and Observations**
  - Likelihoods quantify how well a model explains the data.
- **Customization**
  - Different datasets require tailored likelihood functions.

### Gaussian Likelihood

- **Assumption**
  - Measurement errors are normally distributed.
- **Justification**
  - Central Limit Theorem: The sum of many small, independent random variables tends toward a normal distribution.

### Gaussian Likelihood Formula

$$
\mathcal{L}(\theta) = \prod_{i=1}^N \frac{1}{\sqrt{2\pi\sigma_i^2}} \exp\left( -\frac{(D_i - M_i(\theta))^2}{2\sigma_i^2} \right)
$$

- ( $\theta$ ): Model parameters.
- ( $D_i$ ): Observed data points.
- ( $M_i(\theta)$ ): Model predictions.
- ( $\sigma_i$ ): Measurement uncertainties.

### Simplification for Independent Data

- If ( $\sigma_i$ ) are constants:
  $$
  \mathcal{L} \propto \exp\left( -\frac{1}{2} \chi^2 \right)
  $$
- Where:
  $$
  \chi^2 = \sum_{i=1}^N \frac{(D_i - M_i(\theta))^2}{\sigma_i^2}
  $$

### Challenges in Cosmology

- **Correlated Data**
  - Some datasets have correlated errors.
- **Non-Gaussian Errors**
  - Certain measurements may have asymmetric or non-normal errors.

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import imageio
import os

# Parameters
n_max = 10           # Maximum number of uniform variables to sum
num_samples = 100000 # Number of samples to generate for each n
image_files = []     # List to store the filenames of generated images

# Loop over n from 1 to n_max
for n in range(1, n_max+1):
    # Generate num_samples samples of the sum of n uniform(0,1) random variables
    sums = np.sum(np.random.uniform(0,1,(num_samples,n)), axis=1)
    
    # Compute histogram of the sums
    counts, bin_edges = np.histogram(sums, bins=100, density=True)
    bin_centers = (bin_edges[:-1] + bin_edges[1:])/2

    # Create a new figure
    plt.figure(figsize=(8,6))
    # Plot the histogram as a bar plot
    plt.bar(bin_centers, counts, width=bin_edges[1]-bin_edges[0], alpha=0.6, label='Sum of {} uniforms'.format(n))

    # Compute parameters of the normal distribution approximation
    mu = n * 0.5                      # Mean of the sum of n Uniform(0,1) variables
    sigma = np.sqrt(n/12)             # Standard deviation of the sum of n Uniform(0,1) variables
    # Generate x-values for plotting the normal distribution
    x = np.linspace(min(sums), max(sums), 1000)
    # Plot the normal distribution curve
    plt.plot(x, norm.pdf(x, mu, sigma), 'r-', lw=2, label='Normal approximation')

    # Add title and labels
    plt.title('Distribution of the Sum of {} Uniform(0,1) Variables'.format(n))
    plt.xlabel('Sum')
    plt.ylabel('Probability Density')
    plt.legend()

    # Save the plot to an image file
    filename = 'sum_uniforms_n{}.png'.format(n)
    plt.savefig(filename)
    image_files.append(filename)  # Add filename to list
    plt.close()  # Close the figure to free memory

# Create an animated GIF from the saved images with increased duration and looping
with imageio.get_writer('sum_uniforms.gif', mode='I', duration=1000.0, loop=0) as writer:
    for filename in image_files:
        image = imageio.imread(filename)
        writer.append_data(image)

# Optionally, remove the image files to clean up
for filename in image_files:
    os.remove(filename)
```

## Application to Type Ia Supernovae

### Importance in Cosmology

- **Standardizable Candles**
  - Type Ia supernovae have consistent intrinsic brightness, allowing distance measurements.
- **Measuring Cosmic Expansion**
  - Observations help determine the rate of expansion of the universe.

### Simplistic Likelihood for Supernovae

- **Assumptions**
  - Each supernova observation is independent.
  - Measurement errors are normally distributed.

### Likelihood Formula

$$
\mathcal{L} = \exp\left( -\frac{1}{2} \sum_{i=1}^N \frac{( \mu_{\text{obs}, i} - \mu_{\text{theo}, i} )^2}{\sigma_{\mu,i}^2} \right)
$$

- ( $\mu_{\text{obs}, i}$ ): Observed distance modulus.
- ( $\mu_{\text{theo}, i}$ ): Theoretical distance modulus calculated from the model.
- ( $\sigma_{\mu,i}$ ): Uncertainty in the distance modulus.

### Visualizing the Data

- **Distance Modulus ( $\mu$ ) vs. Redshift ( $z$ )**
  - Plotting ( $\mu$ ) against ( $z$ ) reveals the expansion history.

## Calculating the Distance Modulus

### Definition

- The distance modulus relates the luminosity distance to observed magnitudes.

### Formula

$$
\mu = 5 \log_{10} \left( \frac{d_L}{1\,\text{Mpc}} \right) + 25
$$

- ( $d_L$ ): Luminosity distance in megaparsecs (Mpc).

### Calculating the Luminosity Distance ( $d_L$ )

1. **Friedmann-Robertson-Walker (FRW) Metric**
   - Describes a homogeneous and isotropic universe.

2. **Luminosity Distance Integral**

$$
d_L(z) = (1 + z) \frac{c}{H_0} \int_0^z \frac{dz'}{E(z')}
$$

- ( $c$ ): Speed of light.
- ( $H_0$ ): Hubble constant.
- ( $E(z)$ ): Dimensionless Hubble parameter.

3. **Dimensionless Hubble Parameter**

For a flat universe ( $\Omega_k = 0$ ):

$$
E(z) = \sqrt{ \Omega_m (1 + z)^3 + \Omega_\Lambda }
$$

- ( $\Omega_m$ ): Matter density parameter.
- ( $\Omega_\Lambda$ ): Dark energy density parameter.

## Introduction to Markov Chain Monte Carlo (MCMC)

### Purpose of MCMC

- **Sampling the Posterior**
  - Provides a way to estimate the probability distributions of parameters.
- **High-Dimensional Spaces**
  - Efficiently explores parameter spaces that are difficult to sample directly.

### Basics of Markov Chains

- **Markov Property**
  - The next state depends only on the current state, not on the sequence of events that preceded it.
- **Chain Construction**
  - A sequence of samples where each sample depends on the previous one.

### Monte Carlo Methods

- **Random Sampling**
  - Uses randomness to solve problems that might be deterministic in principle.
- **Estimating Integrals**
  - Approximate multi-dimensional integrals (e.g., Bayesian evidence).

### How MCMC Works

1. **Initialize Parameters**
   - Start at an initial point in parameter space.

2. **Proposal Distribution**
   - Suggest a new point based on the current point.

3. **Acceptance Criterion**
   - Decide whether to accept or reject the new point using the Metropolis-Hastings algorithm.

4. **Iterate**
   - Repeat the process to build up a chain of accepted points.

### Metropolis-Hastings Algorithm

- **Acceptance Probability (\( \alpha \))**

$$
\alpha = \min\left(1, \frac{P(H_{\text{new}}|D)}{P(H_{\text{current}}|D)} \cdot \frac{q(H_{\text{current}}|H_{\text{new}})}{q(H_{\text{new}}|H_{\text{current}})} \right)
$$

- \( q \): Proposal distribution.

### Visualization

- **Trace Plots**
  - Show parameter values over iterations.
- **Corner Plots**
  - Display posterior distributions and parameter correlations.

### Tools

- **Python Libraries**
  - `emcee`: A popular, easy-to-use MCMC sampler.

## Bayesian vs. Frequentist Revisited

### Advantages of Bayesian Methods

- **Incorporation of Prior Knowledge**
  - Ability to include previous results or theoretical considerations.
- **Probabilistic Interpretation**
  - Provides full probability distributions, not just point estimates.
- **Model Comparison**
  - Can compare models with different numbers of parameters using Bayesian evidence.

### Criticisms and Challenges

- **Subjectivity**
  - Choice of priors can influence results.
- **Computational Complexity**
  - MCMC methods can be computationally intensive.

### Frequentist Perspective

- **Objectivity**
  - Relies less on subjective prior choices.
- **Simplicity**
  - Often computationally simpler for parameter estimation.

### Complementary Approaches

- **Use Both Methods**
  - Frequentist methods can be used for initial estimates.
  - Bayesian methods can refine estimates and provide uncertainty quantification.

# Additional Notes on MCMC

### Common MCMC Algorithms

- **Metropolis-Hastings**
  - Simple and widely used, but can be inefficient for complex problems.
- **Gibbs Sampling**
  - Useful when conditional distributions are known.
- **Hamiltonian Monte Carlo (HMC)**
  - Uses gradients to propose states, more efficient for high-dimensional spaces.

### Convergence Diagnostics

- **Burn-in Period**
  - Initial samples may not represent the posterior; discard them.
- **Autocorrelation**
  - Measures how successive samples are related.
- **Effective Sample Size**
  - Accounts for autocorrelation to estimate the number of independent samples.

### Tips for Effective MCMC Sampling

- **Good Initialization**
  - Start walkers in regions of high probability.
- **Tuning the Proposal Distribution**
  - Balance between exploration and acceptance rate.
- **Multiple Chains**
  - Running multiple chains can help assess convergence.
